# Career Trajectory Optimizer — AI Career Coach for Indian Tech Professionals

A hands-on tutorial of the **Deep Agents** framework — LangChain's batteries-included
agent layer that sits on top of LangGraph — built to the exact technical pattern of the
`Deep Agent` GTM tutorial in this workspace.

We build a **career-coach assistant**: an orchestrator that delegates to eight specialist
subagents — including a dedicated web-search researcher — to parse a profile, gather live
Indian tech market data, benchmark salary, track market demand, score risk, and — when the
user states a career goal — build a personalized 30-60-90 day roadmap.

> **Analysis-only.** Nothing is ever actually synced to a calendar — the "sync" step is a
> simulation behind a human-approval gate.

## Architecture at a glance

```
                    ┌─────────────────────────────────────┐
  Profile / Goal     │   Career Optimizer Orchestrator       │
        │            │  (plans, detects intent, delegates)   │
        └───────────▶│  write_todos → task → sync_to_calendar│
                     └───┬──────┬──────┬──────┬──────┬──────┬─┘
                         │task  │task  │task  │task  │task  │task
                    ┌────▼──┐┌──▼───┐┌─▼────┐┌▼─────┐┌▼─────┐┌▼─────┐
                    │profile││salary││market││risk  ││career││roadmap│
                    │analyzr││intel ││track ││scorer││recmdr││planner│
                    └────┬──┘└──┬───┘└─┬────┘└┬─────┘└┬─────┘└┬─────┘
                         └──────┴──────┴──────┴────────┴───────┘
                                        │  (+ critic reviews all)
                              ┌─────────▼─────────┐
                              │   SHARED STATE    │
                              │ files + todos     │  ← findings pass between agents here
                              └───────────────────┘
   Persistence:  disk (/profile, /analysis, /roadmap)  +  Store (/memories — cross-session)
```

**Deep Agents concepts this notebook demonstrates:**

| Concept | Where you'll see it |
|---|---|
| **Planning** | `write_todos` breaks the request into steps |
| **Subagents** | `task` delegates to specialists, each with its own tools, prompt, and *isolated context* |
| **Shared state** | message histories are isolated, but the **filesystem + todos are shared** — findings pass between agents via files |
| **Skills** | `SKILL.md` methodology guides loaded on demand (progressive disclosure) |
| **Memory** | `/memories/` persists across sessions via a `Store` |
| **Human-in-the-loop** | the calendar-sync action pauses for your approval |
| **Runaway protection** | `recursion_limit` is the hard backstop on the agent loop |
| **Intent branching** | Profile Analysis Mode vs. Career Aspiration Mode |

## 3. Tools

Three kinds of tools, each teaching a distinct idea:

- **`web_search` (Tavily)** — the agent reaching *outside itself* for fresh salary/market data. Optional.
- **`roadmap_validator` / `risk_dashboard_validator`** — deterministic validators. Teaching point:
  the LLM **drafts**, but plain code **validates** what models are unreliable at
  (checking required sections exist, risk scores are in range).
- **`sync_to_calendar`** — a **simulated** outbound action, wrapped later in a human-approval gate.

![Diagram](./images/03_img.png)

````python
from langchain.tools import tool

# --- Tool 1: Web search (external API) — optional ---------------------
# LinkUp is the PRIMARY provider; Tavily is the fallback if LinkUp isn't
# configured (or fails to initialize) — same optional pattern either way.
research_tools = []
if os.environ.get("LINKUP_API_KEY", "").strip():
    try:
        from langchain_linkup import LinkupSearchTool
        research_tools = [LinkupSearchTool(depth="standard", output_type="searchResults", max_results=3)]
        print("Web search: LinkUp ENABLED (primary)")
    except Exception as e:
        print(f"Web search: LinkUp failed to initialize ({e}) — falling back to Tavily")

if not research_tools and os.environ.get("TAVILY_API_KEY", "").strip():
    from langchain_tavily import TavilySearch
    research_tools = [TavilySearch(max_results=3)]
    print("Web search: Tavily ENABLED (fallback)")

if not research_tools:
    print("Web search: disabled (no LinkUp or Tavily key) — subagents will rely on grounding")

# --- Tools 2 & 3: Deterministic validators -----------------------------
@tool
def roadmap_validator(text: str) -> str:
    """Validate a career roadmap (required sections present). Returns a pass/fail report."""
    required = [
        "goal", "gap", "upskill", "30", "60", "90", "timeline", "salary", "probability",
    ]
    lowered = text.lower()
    missing = [word for word in required if word not in lowered]
    return "PASS" if not missing else "ISSUES:\n- Missing sections/keywords: " + ", ".join(missing)

@tool
def risk_dashboard_validator(salary_risk: int, skill_risk: int, security_risk: int,
                              burnout_risk: int, growth_risk: int) -> str:
    """Validate a risk dashboard: all five risk scores must be present and within 0-10."""
    scores = {
        "salary_risk": salary_risk,
        "skill_risk": skill_risk,
        "security_risk": security_risk,
        "burnout_risk": burnout_risk,
        "growth_risk": growth_risk,
    }
    issues = [f"{name}={value} out of range 0-10" for name, value in scores.items() if not (0 <= value <= 10)]
    return "PASS" if not issues else "ISSUES:\n- " + "\n- ".join(issues)

# --- Tool 4: Mock calendar sync (SIMULATION ONLY) -----------------------
@tool
def sync_to_calendar(task_description: str, cadence: str) -> str:
    """Sync a roadmap task to a calendar (SIMULATED — nothing is actually synced). cadence: 'daily', 'weekly', 'monthly', or 'quarterly'."""
    return f"[SIMULATED] Synced '{task_description}' ({cadence}) to calendar. Nothing was actually synced."

print("Tools ready.")
````

## 5. Backends: files + cross-session memory

Two persistence layers, wired with a **`CompositeBackend`** that routes by path:

- **`/memories/*`** → `StoreBackend` — survives across sessions (career progress we learn once)
- **everything else** → `FilesystemBackend` on real disk — so `/profile`, `/analysis`, and
  `/roadmap` show up in your file explorer during the demo

The **checkpointer** holds per-thread state and is *required* for the human-in-the-loop gate.

![Diagram](./images/05_img.png)

````python
from deepagents.backends import CompositeBackend, FilesystemBackend, StoreBackend
from langgraph.store.memory import InMemoryStore
from langgraph.checkpoint.memory import MemorySaver

store = InMemoryStore()        # cross-session memory (use a PostgresStore in production)
checkpointer = MemorySaver()   # per-thread state + required for HITL interrupts

# resolves the store/runtime itself via get_store(); the `store=` we hand to
# create_deep_agent below is what it picks up. `namespace` sets where memories
backend = CompositeBackend(
    default=FilesystemBackend(root_dir=".", virtual_mode=True),
    routes={"/memories/": StoreBackend(namespace=lambda ctx: ("career",))},
)
print("Backend: disk (default) + Store (/memories/) | checkpointer + store ready")
````

## 6. Specialist subagents

Each subagent is a dict: a name, a description (the orchestrator reads this to decide *when*
to delegate), a focused system prompt, its own tools, and its own skills. Subagents are
**stateless** — each `task` call is a fresh run — so instructions must be complete.

Note the **division of labour**: only `market-researcher` holds the `web_search` (Tavily)
tool — every other analysis subagent reads its findings from `/research/market_brief.md`
instead of searching independently. The recommender and planner run only in Career
Aspiration Mode; the rest run in both modes. Clean tools = clean context.

![Diagram](./images/06_img.png)

````python
SKILLS = ["/skills/"]  # loaded from disk via the FilesystemBackend

profile_analyzer_subagent = {
    "name": "profile-analyzer",
    "description": "Parses the raw profile/resume text into a structured profile and builds a career timeline. Use first, always.",
    "system_prompt": (
        "You parse a raw career profile (resume text or manual entry). Extract role, company, "
        "years of experience, skills, and education. Detect education tier (IIT/NIT = premium "
        "tier per grounding). Build a timeline of past roles and flag job-hopping or long tenure. "
        "Write the structured profile and timeline to /profile/timeline.md and return a one-line summary."
    ),
    "tools": [],
}

market_researcher_subagent = {
    "name": "market-researcher",
    "description": "Gathers live Indian tech market data (salary trends, hiring demand, GenAI growth) via web search. Use second, right after profile-analyzer.",
    "system_prompt": (
        "Research the current Indian tech job market using web_search if available: salary "
        "benchmarks for the candidate's role/experience/location, hiring demand for their "
        "skills, and GenAI/emerging-skill growth trends. If no web_search tool is available, "
        "rely on the grounding facts only and say so explicitly. Write 6-10 crisp, sourced "
        "bullet points to /research/market_brief.md and return a one-line summary."
    ),
    "tools": research_tools,
}

salary_intelligence_subagent = {
    "name": "salary-intelligence",
    "description": "Benchmarks the candidate's current pay against the Indian market. Use after market-researcher.",
    "system_prompt": (
        "Load the 'salary-benchmarking' skill and follow it strictly. Read /profile/timeline.md "
        "and /research/market_brief.md. Ground every number in those two files only — do not "
        "search the web yourself. Produce a market range, the candidate's estimated position in "
        "it, and (if a goal salary was given in the brief) the gap and levers to close it. "
        "Save the result to /analysis/salary.md and return a one-line summary."
    ),
    "tools": [],
    "skills": SKILLS,
}

market_tracker_subagent = {
    "name": "market-tracker",
    "description": "Tracks skill demand and job-market trends relevant to the candidate's skills. Use after market-researcher.",
    "system_prompt": (
        "Read /profile/timeline.md and /research/market_brief.md. Do not search the web "
        "yourself — reason only from those two files plus the grounding facts (e.g. GenAI "
        "~180% YoY growth). Identify which of the candidate's skills are growing, flat, or "
        "declining, and which emerging skills are worth learning. Save to "
        "/analysis/market_trends.md and return a one-line summary."
    ),
    "tools": [],
}

risk_scorer_subagent = {
    "name": "risk-scorer",
    "description": "Scores readiness for the next level and produces a 5-axis risk dashboard. Use after salary-intelligence and market-tracker.",
    "system_prompt": (
        "Read /profile/timeline.md, /analysis/salary.md, and /analysis/market_trends.md. Score "
        "readiness for the next career level from the candidate's accomplishments. Then score "
        "five risks 0-10 each: salary_risk, skill_risk, security_risk, burnout_risk, growth_risk. "
        "Run risk_dashboard_validator on your five scores and fix any issues it reports. Save the "
        "full dashboard to /analysis/risk_dashboard.md and return a one-line summary."
    ),
    "tools": [risk_dashboard_validator],
}

career_recommender_subagent = {
    "name": "career-recommender",
    "description": (
        "Builds goal-based recommendations for Career Aspiration Mode only (a stated salary, "
        "company, or role goal). Use after risk-scorer, only when the user gave a goal."
    ),
    "system_prompt": (
        "Read /profile/timeline.md, /research/market_brief.md, /analysis/salary.md, "
        "/analysis/market_trends.md, and /analysis/risk_dashboard.md. Do not search the web "
        "yourself — use only those files. The user's goal is embedded in the task you were "
        "given. Apply this logic: goal salary -> identify target company tier and required "
        "upskills; goal company (e.g. Microsoft) -> use /research/market_brief.md for the "
        "company's bar and gap vs. candidate; goal role (e.g. Tech Lead) -> identify "
        "leadership/architecture gaps and visibility actions; goal 'GenAI Engineer' -> note the "
        "~20-30% skill premium and a 3-6 month learning plan; goal US remote/relocation -> note "
        "the ~20-30% feasibility and 12-18 month prep. Save the gap analysis and recommendation "
        "to /analysis/recommendation.md and return a one-line summary."
    ),
    "tools": [],
}

roadmap_planner_subagent = {
    "name": "roadmap-planner",
    "description": "Turns the recommendation into a 30-60-90 day roadmap. Use only in Career Aspiration Mode, after career-recommender.",
    "system_prompt": (
        "Load the 'roadmap-planning' skill and follow it strictly. Read /analysis/recommendation.md. "
        "Produce: goal summary, gap analysis, upskilling plan, a 30-60-90 day plan, timeline to "
        "goal, salary projection, and a success probability percentage. Run roadmap_validator on "
        "your draft and fix any issues it reports. Save the roadmap to /roadmap/plan.md and return "
        "the final roadmap text."
    ),
    "tools": [roadmap_validator],
    "skills": SKILLS,
}

critic_subagent = {
    "name": "critic",
    "description": (
        "Reviews all analysis files for grounding accuracy, methodology compliance, and quality; "
        "runs the validators; suggests fixes."
    ),
    "system_prompt": (
        "Read every file that exists under /profile, /research, /analysis, and /roadmap. Review "
        "on four axes:\n"
        "1. GROUNDING ACCURACY — flag any number or claim not traceable to the grounding file or "
        "/research/market_brief.md.\n"
        "2. METHODOLOGY COMPLIANCE — load the 'salary-benchmarking' and 'roadmap-planning' skills "
        "and check the relevant files actually follow the methodology.\n"
        "3. INTERNAL CONSISTENCY — do the salary, risk, and roadmap files agree with each other?\n"
        "4. VALIDATORS — run roadmap_validator (if a roadmap exists) and risk_dashboard_validator "
        "(if a risk dashboard exists).\n"
        "Return a short review with concrete, actionable fixes, calling out any unsupported "
        "numbers or skill violations explicitly."
    ),
    "tools": [roadmap_validator, risk_dashboard_validator],
    "skills": SKILLS,
}

print("8 subagents defined.")
````

## 7. Assemble the deep agent

This is the whole point of Deep Agents: you **configure**, you don't build the plumbing. Planning, the filesystem tools, and delegation come for free.

![Diagram](./images/07_img.png)

````python
from deepagents import create_deep_agent

ORCHESTRATOR_PROMPT = f"""You are an AI career coach for Indian tech professionals.
For a profile you produce either a market-position analysis or a full goal-based roadmap.

Market grounding (always honor this; never invent facts beyond it + research):
{GROUNDING}

Workflow — first call write_todos to plan, then:
1. Delegate to 'profile-analyzer' to parse the profile (writes /profile/timeline.md).
2. Delegate to 'market-researcher' to gather live Indian tech market data via web search
   (writes /research/market_brief.md).
3. Delegate to 'salary-intelligence' (writes /analysis/salary.md).
4. Delegate to 'market-tracker' (writes /analysis/market_trends.md).
5. Delegate to 'risk-scorer' (writes /analysis/risk_dashboard.md).
6. INTENT DETECTION: read the user's brief.
   - If it states a concrete goal (a target salary, company, or role/timeline), this is
     CAREER ASPIRATION MODE: delegate to 'career-recommender' (writes
     /analysis/recommendation.md), then 'roadmap-planner' (writes /roadmap/plan.md).
   - Otherwise this is PROFILE ANALYSIS MODE: skip step 6's subagents and go straight to
     step 7 with only the salary/market/risk analysis.
7. Delegate to 'critic' to review everything produced so far; if it flags something
   important, re-delegate to fix it.
8. Save a one-paragraph progress note (mode used, key findings, current goal if any) to
   /memories/career_progress.md for future coaching sessions.
9. ONLY in Career Aspiration Mode: sync the roadmap's monthly checkpoints to a calendar.
   Emit one sync_to_calendar call per checkpoint (cadence='monthly' for month-end reviews,
   cadence='weekly' for milestones). Do not write a final summary until all sync_to_calendar
   calls have been made and returned. In Profile Analysis Mode, skip this step entirely.

Keep everything concise, data-grounded, and honest about timelines."""

agent = create_deep_agent(
    model=MODEL,
    system_prompt=ORCHESTRATOR_PROMPT,
    tools=[sync_to_calendar],
    subagents=[
        profile_analyzer_subagent,
        market_researcher_subagent,
        salary_intelligence_subagent,
        market_tracker_subagent,
        risk_scorer_subagent,
        career_recommender_subagent,
        roadmap_planner_subagent,
        critic_subagent,
    ],
    backend=backend,
    store=store,
    skills=SKILLS,
    interrupt_on={"sync_to_calendar": True},   # HITL gate — needs the checkpointer
    checkpointer=checkpointer,
)
print("Deep agent assembled.")
````

## 8. Run a profile — Profile Analysis Mode

We pass a profile with no explicit goal, so the orchestrator should stay in Profile Analysis
Mode and skip the recommender/planner/calendar steps. Watch the message stream: you'll see
the orchestrator's `task` calls and each subagent's **final report** — but *not* the
subagents' internal back-and-forth. That hidden churn is the context-isolation benefit.

`recursion_limit` caps the agent loop — the hard backstop against runaway loops.

![Diagram](./images/08_img.png)

````python
config = {"configurable": {"thread_id": "career-demo-1"}, "recursion_limit": 60}

profile_brief = (
    "Profile: 4 years as a Backend Engineer at a mid-size product company in Bengaluru, "
    "B.Tech from a tier-2 college, skills in Python, Django, PostgreSQL, AWS. No specific "
    "goal yet — just show me where I stand in the market today."
)

result = agent.invoke({"messages": [{"role": "user", "content": profile_brief}]}, config=config)
# try streaming the output to see how the agent is executing

for m in result["messages"]:
    m.pretty_print()
````

## 8b. Run with a stated goal — Career Aspiration Mode

Same agent, a new thread, but this brief states a concrete goal. The orchestrator should now
take the Career Aspiration Mode branch: delegate to `career-recommender` and `roadmap-planner`,
and finish by proposing `sync_to_calendar` calls for the roadmap's checkpoints.

````python
config_goal = {"configurable": {"thread_id": "career-demo-2"}, "recursion_limit": 60}

goal_brief = (
    "Profile: 4 years as a Backend Engineer at a mid-size product company in Bengaluru, "
    "B.Tech from a tier-2 college, skills in Python, Django, PostgreSQL, AWS. "
    "Goal: I want to earn ₹100L within 2 years, ideally as a GenAI Engineer."
)

result = agent.invoke({"messages": [{"role": "user", "content": goal_brief}]}, config=config_goal)

for m in result["messages"]:
    m.pretty_print()
````

## 9. Human-in-the-loop: approve the calendar-sync gate

We set `interrupt_on={"sync_to_calendar": True}`, so the run **pauses** before each sync and
hands control back to us. We inspect what it wants to do, then approve (or reject with
feedback, or edit the arguments) and resume with a `Command`.

![Diagram](./images/09_img.png)

````python
from langgraph.types import Command

# What is the agent waiting to do? In this LangChain version each interrupt's
# .value is a SINGLE HITLRequest dict: {"action_requests": [...], "review_configs": [...]}.
# One action_request == one hanging tool call, and we must send one decision PER
# action_request — not one per interrupt.
#
# How MANY action_requests appear depends on what the model emitted: if it issues
# several sync_to_calendar calls in one step, they batch into a single request with
# multiple action_requests. If it only emits one, you'll see one here, approve it,
# resume — and the next call surfaces as the next interrupt on the following loop
# iteration. The while-loop below handles either case.
def pending_requests(res):
    reqs = []
    for intr in res.get("__interrupt__", []):
        val = intr.value
        if isinstance(val, dict) and "action_requests" in val:
            reqs.extend(val["action_requests"])      # unwrap the HITLRequest
        elif isinstance(val, list):
            reqs.extend(val)                          # older list-style payload
        else:
            reqs.append(val)
    return reqs

reqs = pending_requests(result)
if reqs:
    for r in reqs:
        print("PENDING APPROVAL:", r)
else:
    print("No interrupt pending (agent may have finished without calling sync_to_calendar).")

# Approve each pending sync and resume. One decision PER hanging tool call.
# Loops until no interrupt remains — so it catches sync calls that arrive across
# multiple turns, not just a single batched request.
# To reject instead, use: {"type": "reject", "message": "space these out more"}
while "__interrupt__" in result:
    n = len(pending_requests(result))
    decisions = [{"type": "approve"} for _ in range(n)]
    result = agent.invoke(Command(resume={"decisions": decisions}), config=config_goal)

# Sanity check: confirm the calendar-sync calls were actually made.
synced = [
    tc["args"].get("task_description")
    for m in result["messages"]
    for tc in (getattr(m, "tool_calls", None) or [])
    if tc["name"] == "sync_to_calendar"
]
print("\nSynced checkpoints:", synced)
if not synced:
    print("  WARNING: sync_to_calendar was never called — the model may have skipped it.")

print("\n===== FINAL RESPONSE =====")
result["messages"][-1].pretty_print()
````

## 10. Persistent memory across sessions

A **new thread** = a new session with a fresh message history. But anything written to
`/memories/` lives in the `Store`, not the thread — so a brand-new conversation (e.g. next
month's check-in) can still recall the candidate's goal and prior findings. This is the
difference between *thread state* and *long-term memory*.

![Diagram](./images/10_img.png)

````python
config_checkin = {"configurable": {"thread_id": "career-demo-3"}, "recursion_limit": 20}
result_checkin = agent.invoke(
    {"messages": [{"role": "user",
                   "content": "Read /memories/career_progress.md and summarise where we left off in one line."}]},
    config=config_checkin,
)
result_checkin["messages"][-1].pretty_print()
````

## Recap

You built a multi-agent career-coaching system with almost no plumbing code:

- An **orchestrator** that *planned* (`write_todos`), *branched on intent* (Profile Analysis
  vs. Career Aspiration Mode), and *delegated* (`task`).
- Eight **subagents** with isolated context and focused tools, including a dedicated
  `market-researcher` that is the *only* subagent holding the `web_search` tool.
- **Skills** for benchmarking and roadmap methodology, loaded on demand.
- A **composite backend**: disk for artifacts, a Store for cross-session memory.
- A **human-in-the-loop** gate on the (simulated) calendar-sync step.
- `recursion_limit` as the safety backstop.

**Where to take it next:** swap `InMemoryStore` for a `PostgresStore`, add a real resume-parsing
tool (Claude Vision API) ahead of `profile-analyzer`, wire `sync_to_calendar` to the real Google
Calendar API behind the same HITL gate, or give the critic a structured `response_format` so its
verdict is machine-readable.

## Interaction flow

```mermaid
sequenceDiagram

    N->>F: agent.invoke(profile / goal brief)
    F->>L: Send prompt, tools, and subagent definitions
    L->>F: write_todos(plan)
    F-->>L: Store plan in agent state

    L->>F: task(profile-analyzer)
    F->>S: Send profile-parsing prompt
    S->>B: Write /profile/timeline.md
    S-->>L: Return profile summary

    L->>F: task(market-researcher)
    F->>S: Send market-research prompt (only subagent with web_search)
    S->>T: Search web if Tavily is enabled
    T-->>S: Return search results
    S->>B: Write /research/market_brief.md
    S-->>L: Return market-research summary

    L->>F: task(salary-intelligence)
    F->>S: Send benchmarking prompt and skill
    S->>B: Read /research/market_brief.md
    S->>B: Write /analysis/salary.md
    S-->>L: Return salary summary

    L->>F: task(market-tracker)
    F->>S: Send demand-tracking prompt
    S->>B: Read /research/market_brief.md
    S->>B: Write /analysis/market_trends.md
    S-->>L: Return trends summary

    L->>F: task(risk-scorer)
    F->>S: Send risk-scoring prompt and validator
    S->>F: Run risk_dashboard_validator
    F-->>S: Return PASS or issues
    S->>B: Write /analysis/risk_dashboard.md
    S-->>L: Return risk summary

    alt Career Aspiration Mode (goal stated)
        L->>F: task(career-recommender)
        F->>S: Send goal-based recommendation prompt
        S->>B: Read /research/market_brief.md
        S->>B: Write /analysis/recommendation.md
        S-->>L: Return recommendation

        L->>F: task(roadmap-planner)
        F->>S: Send roadmap prompt and validator
        S->>F: Run roadmap_validator
        F-->>S: Return PASS or issues
        S->>B: Write /roadmap/plan.md
        S-->>L: Return roadmap
    end

    L->>F: task(critic)
    F->>S: Send review prompt and all files
    S->>B: Read profile, research, analysis, and roadmap files
    S-->>L: Return review and fixes

    L->>B: Write /memories/career_progress.md

    opt Career Aspiration Mode only
        L->>F: Call sync_to_calendar for each checkpoint
        F-->>H: Pause for approval
        H->>F: Approve sync calls
        F-->>L: Return sync results
    end

    L-->>N: Return final result
```
